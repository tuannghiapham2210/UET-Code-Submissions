## Bài 18
**Đề bài:** Dùng Python đọc bộ dữ liệu mô phỏng `ball_bearing` trên OpenIntro. Mỗi hàng ghi tuổi thọ `life_span` của một ổ bi, tính bằng giờ. Tải tệp CSV từ trang dữ liệu.
- **a.** Tính cỡ mẫu, trung bình mẫu và độ lệch chuẩn mẫu của `life_span`.
- **b.** Lập và diễn giải khoảng tin cậy 95% cho tuổi thọ trung bình.


```python
# Đọc dữ liệu
x = pd.read_csv("../data/ball_bearing.csv")["life_span"].dropna()

# a. Cỡ mẫu, trung bình mẫu và độ lệch chuẩn mẫu
n = len(x)
x_bar = x.mean()
s = x.std(ddof=1)

# b. Khoảng tin cậy 95% cho tuổi thọ trung bình
se = s / np.sqrt(n)
ci_t = t.interval(0.95, df=n-1, loc=x_bar, scale=se)
ci_z = norm.interval(0.95, loc=x_bar, scale=se)

print(f"a) Cỡ mẫu n = {n}")
print(f"   Trung bình mẫu x̄ = {x_bar:.4f} giờ")
print(f"   Độ lệch chuẩn mẫu s = {s:.4f} giờ")
print(f"b) Khoảng tin cậy 95% (phân phối t): [{ci_t[0]:.4f}, {ci_t[1]:.4f}] giờ")
print(f"   Khoảng tin cậy 95% (xấp xỉ chuẩn z): [{ci_z[0]:.4f}, {ci_z[1]:.4f}] giờ")
```

**Kết quả & Diễn giải (Bài 18):**
- **a.** Cỡ mẫu $n = 75$, trung bình mẫu $\bar{x} \approx 6.8492$ giờ, độ lệch chuẩn mẫu $s \approx 1.2510$ giờ.
- **b.** Sai số chuẩn $SE = \dfrac{s}{\sqrt{n}} \approx 0.1445$ giờ.
  - Khoảng tin cậy $95\%$ theo phân phối $t$ ($df = 74$): $[6.5614;\, 7.1370]$ giờ (hoặc theo xấp xỉ chuẩn $z$: $[6.5661;\, 7.1323]$ giờ).
  - **Diễn giải:** Ta tin cậy $95\%$ rằng tuổi thọ trung bình thực sự $\mu$ của tổng thể các ổ bi loại này nằm trong khoảng từ $6.5614$ giờ đến $7.1370$ giờ.

## Bài 20
**Đề bài:** Dùng Python đọc bộ dữ liệu `acs12` trên OpenIntro, trích từ Khảo sát cộng đồng Hoa Kỳ năm 2012. Mỗi hàng ghi thông tin của một người tham gia khảo sát; cột `disability` cho biết người đó có tình trạng khuyết tật hay không. Tải tệp CSV từ trang dữ liệu.
- **a.** Tính số câu trả lời hợp lệ và tỷ lệ người có `disability` bằng `yes`.
- **b.** Lập và diễn giải khoảng tin cậy 95% cho tỷ lệ này.


```python
# Đọc dữ liệu
d = pd.read_csv("../data/acs12.csv")["disability"].dropna()

# a. Số câu trả lời hợp lệ và tỷ lệ mẫu
n = len(d)
k = (d == "yes").sum()
p_hat = k / n

# b. Khoảng tin cậy 95% cho tỷ lệ tổng thể
se = np.sqrt(p_hat * (1 - p_hat) / n)
ci = norm.interval(0.95, loc=p_hat, scale=se)

print(f"a) Số câu trả lời hợp lệ n = {n} (số người 'yes' = {k})")
print(f"   Tỷ lệ mẫu p̂ = {p_hat:.4f} ({p_hat*100:.2f}%)")
print(f"b) Khoảng tin cậy 95%: [{ci[0]:.4f}, {ci[1]:.4f}] (tức {ci[0]*100:.2f}% đến {ci[1]*100:.2f}%)")
```

**Kết quả & Diễn giải (Bài 20):**
- **a.** Số câu trả lời hợp lệ $n = 2000$ (trong đó có $k = 324$ người trả lời `yes`). Tỷ lệ mẫu $\hat{p} = \dfrac{324}{2000} = 0.162$ ($16.2\%$).
- **b.** Kiểm tra điều kiện xấp xỉ chuẩn: mẫu ngẫu nhiên cỡ $n = 2000$ nhỏ hơn $10\%$ dân số Hoa Kỳ, đồng thời $n\hat{p} = 324 \ge 10$ và $n(1 - \hat{p}) = 1676 \ge 10$ (thỏa mãn).
  - Sai số chuẩn $\widehat{SE}(\hat{p}) = \sqrt{\dfrac{\hat{p}(1 - \hat{p})}{n}} \approx 0.00824$.
  - Khoảng tin cậy $95\%$: $[0.1459;\, 0.1781]$ (hay $[14.59\%;\, 17.81\%]$).
  - **Diễn giải:** Ta tin cậy $95\%$ rằng tỷ lệ người có tình trạng khuyết tật trong toàn bộ dân số Hoa Kỳ năm 2012 nằm trong khoảng từ $14.59\%$ đến $17.81\%$.
